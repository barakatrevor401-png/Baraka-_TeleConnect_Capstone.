 ## STEP 5 :  STATISTICAL SUMMARY

 1. MEAN

```python
total_charges_mean = df["total_charges"].mean()
total_charges_median = df["total_charges"].median()

q3 = df["total_charges"].quantile(0.75) # upper quantile
q1 = df["total_charges"].quantile(0.25) #lower quantile
iqr = q3-q1 # interquantile range
 #boundaries
boundary_1 = q1 - 1.5 * iqr
boundary_2 = q3 + 1.5 *iqr

#checking for outliers
outliers =df[(df["total_charges"] < boundary_1) | (df["total_charges"] >boundary_2)]

#checking for skewness
skew = df["total_charges"].skew()

#ouput code
print("total charges mean :", total_charges_mean)
print("total charges median:", total_charges_median)
print("upper quantile : " , q3 , " ,","lower quantile :" , q1 )
print(len(outliers))
print("skewness:",skew)
```


From the numerical summary of the total charges, the dataset seems right skewed since the score is greater than 1 implying that many customers have low total charges

It also has 105 outliers showing some customers have very high total_charges depending on the services used and the tenure months


#PHASE 4 :DATA VISUALISATION

```python
#histogram showing the skewness of total charges
plt.figure(figsize=(7,6))
sns.histplot(  x = df["total_charges"] , color = "blue" ,)
plt.xlabel("TOTAL CHARGES")
plt.ylabel("COUNT")
plt.title("TOTAL CHARGES DISTRIBUTION")
plt.show()
```

From the histogram ,it clearly shows how the data is right skewed since the  peak is at the far left and it gets progressively shorter as it moves to the right .

This shows majority of the customers are fixated on the low and middle total charges

```python

#create the scatter plot
plt.figure(figsize=(12, 6))
sns.scatterplot( x="tenure_months",y="total_charges" , data=df ,color ="blue")
plt.xlabel("Tenure (Months)")#Label  the horizontal axis line
plt.ylabel("Total Charges")# label the  vertical axis line
plt.title(" Relationship  between Total charges  and tenure ", fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

The scatter plot represents relationship between  tenure months and total charges.

This shows  a positive correlation between total charges and tenure months. The lower the tenure months the lower the charges while the higher the tenure months the higher the charges.

This indicates that if customer has lower enure months, the total charges are low .However if the tenure months are high,the total charge becomes high
